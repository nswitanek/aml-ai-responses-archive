```python
import mltable
from mltable import MLTableHeaders, MLTableFileEncoding, DataType
from azure.ai.ml import MLClient
from azure.ai.ml.entities import Data
from azure.ai.ml.constants import AssetTypes
from azure.identity import DefaultAzureCredential

```

```python
# create paths to the data files
paths = [{"file": "./data/imm_lowest_cost_silver_premiums.csv"}]

# create an MLTable from the data files
tbl = mltable.from_delimited_files(
    paths=paths,
    delimiter=",",
    header=MLTableHeaders.all_files_same_headers,
    infer_column_types=False,
    include_path_column=False,
    encoding=MLTableFileEncoding.utf8,
)
```

```python
# Set the column types.
# We can use the DataType.to_*() functions to convert the column types
column_types = {
    "plan year": DataType.to_int(),
    "State Code": DataType.to_string(),
    "FIPS County Code": DataType.to_string(),
    "County Name": DataType.to_string(),
    "Issuer Name": DataType.to_string(),
    "HIOS Issuer ID": DataType.to_string(),
    "Rating Area": DataType.to_string(),
    "premium": DataType.to_float(),
    "premium_lag": DataType.to_float(),
    "premium_change": DataType.to_float(),
}
tbl = tbl.convert_column_types(column_types)
```

```python
tbl.show(5)
```

```python
tbl = tbl.drop_columns(["premium"]) # keep only premium_change (from last year), or premium (this year premium)
```

```python
# save the MLTable to local disk - in the same directory as the titanic.csv file
tbl.save("./data")
```
