```python
ad = pd.read_excel('Wave 1 Data.xlsx', sheet_name = "AdviceQ")
fa = pd.read_excel('Wave 1 Data.xlsx', sheet_name = "FactQ")
pe = pd.read_excel('Wave 1 Data.xlsx', sheet_name = "PersonalityQ")
so = pd.read_excel('Wave 1 Data.xlsx', sheet_name = "SocialQ")
```

```python
newIDs, questions = [], []
for df in [op, ad, fa, pe, so]: 
    newIDs.extend(df['newID'].values)
    questions.extend(df.iloc[:,3].values)
print(newIDs[:10])
print(newIDs[-10:])
assert len(newIDs) == len(set(newIDs)) # confirm unique newIDs
print(questions[:10])
print(questions[-10:])
```

```python
# put all questions with newIDs in single list
all = pd.DataFrame(data={
    'newID':newIDs,
    'question':questions,
})
```

```python
for i in range(all.shape[0]):
    if i < 10:
        print(f"{all['newID'][i]}: {all['question'][i]}")
```

```python
all.to_csv('all_questions.csv', index=False)
```
