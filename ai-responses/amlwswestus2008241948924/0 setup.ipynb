First, we will
- load all prompts across the five categories

Then, there are models available via the following:
- Hugging Face: Facebook's OPT, Facebook's Galactica model, and OpenAI's GPT2
- OpenAI: GPT-3.5 and ChatGPT
- Microsoft: Turing ULG v2

For Hugging Face models, we will 
- load model
- submit prompts
- collect responses
- write to file.

For OpenAI models, we will 
- submit prompts as requests to models via API
- collect responses
- write to file.

For Microsoft, will likely 
- execute batch inferencing pipeline 
- collect responses 
- write to file.

Then, we can
- compile all responses across models

If setting up environment, in terminal run the following:
- conda create -n py38_PT python=3.8
- conda activate py38_PT
- pip install pytorch torchvision torchaudio --extra-index-url https://download.pytorch.org/whl/cu114  # if using GPU
- pip install accelerate # if using GPU
- pip install transformers\[torch\] # if using CPU
- pip install datasets transformers\[sentencepiece\]
- pip install jupyter ipykernel
- python -m ipykernel install --user --name py38_PT --display-name "Python 3.8 PyTorch"
